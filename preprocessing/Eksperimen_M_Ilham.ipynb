# **5. Data Preprocessing**

Pada tahap ini, data preprocessing adalah langkah penting untuk memastikan kualitas data sebelum digunakan dalam model machine learning.

Jika Anda menggunakan data teks, data mentah sering kali mengandung nilai kosong, duplikasi, atau rentang nilai yang tidak konsisten, yang dapat memengaruhi kinerja model. Oleh karena itu, proses ini bertujuan untuk membersihkan dan mempersiapkan data agar analisis berjalan optimal.

Berikut adalah tahapan-tahapan yang bisa dilakukan, tetapi **tidak terbatas** pada:
1. Menghapus atau Menangani Data Kosong (Missing Values)
2. Menghapus Data Duplikat
3. Normalisasi atau Standarisasi Fitur
4. Deteksi dan Penanganan Outlier
5. Encoding Data Kategorikal
6. Binning (Pengelompokan Data)

Cukup sesuaikan dengan karakteristik data yang kamu gunakan yah. Khususnya ketika kami menggunakan data tidak terstruktur.

```python
print("\n" + "="*50)
print("PREPROCESSING")
print("="*50)

# Menghapus Kolom Tidak Relevan
# 'id' adalah identifier, 'dataset' asal data (mungkin sumber) - tidak digunakan
df_clean = dataset.drop(columns=['id', 'dataset'], errors='ignore')
print("Kolom setelah drop id & dataset:", df_clean.columns.tolist())

# Penanganan Missing Values
# Jika ternyata ada (mungkin di kolom 'ca' atau 'thal'), kita isi dengan median/modus
# Cek missing
missing_cols = df_clean.columns[df_clean.isnull().any()].tolist()
if missing_cols:
    for col in missing_cols:
        if df_clean[col].dtype in ['float64', 'int64']:
            # Isi dengan median
            median_val = df_clean[col].median()
            df_clean[col].fillna(median_val, inplace=True)
            print(f"Kolom {col}: missing diisi median ({median_val})")
        else:
            # Kategorikal diisi modus
            mode_val = df_clean[col].mode()[0]
            df_clean[col].fillna(mode_val, inplace=True)
            print(f"Kolom {col}: missing diisi modus ({mode_val})")
else:
    print("Tidak ada missing values setelah drop kolom.")

# Encoding Data Kategorikal (jika ada)
# Periksa kolom non-numerik
cat_cols = df_clean.select_dtypes(include=['object']).columns.tolist()
print("Kolom kategorikal:", cat_cols)
if cat_cols:
    # Contoh: jika ada kolom 'sex' bertipe object, label encoding atau one-hot
    from sklearn.preprocessing import LabelEncoder
    le = LabelEncoder()
    for col in cat_cols:
        df_clean[col] = le.fit_transform(df_clean[col].astype(str))
    print("Encoding kategorikal selesai.")
else:
    print("Tidak ada kolom kategorikal, semua numerik.")

# Deteksi dan Penanganan Outlier
# Kita gunakan metode IQR pada fitur numerik (selain target)
# Untuk tingkat keparahan, kita akan replace outlier dengan batas (winsorization)
outlier_cols = ['trestbps', 'chol', 'thalch', 'oldpeak']
for col in outlier_cols:
    Q1 = df_clean[col].quantile(0.25)
    Q3 = df_clean[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    # Ganti outlier dengan batas
    df_clean[col] = np.where(df_clean[col] < lower_bound, lower_bound, df_clean[col])
    df_clean[col] = np.where(df_clean[col] > upper_bound, upper_bound, df_clean[col])
    print(f"Outlier {col}: lower={lower_bound:.2f}, upper={upper_bound:.2f} -> winsorized.")

# Normalisasi/Standarisasi Fitur Numerik
from sklearn.preprocessing import StandardScaler

# Pisahkan fitur dan target
X = df_clean.drop(columns=['num'])
y = df_clean['num']

# Pilih kolom untuk scaling (seluruh fitur numerik)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

print("Standarisasi selesai. Shape X_scaled:", X_scaled_df.shape)

# Gabungkan kembali dengan target
df_preprocessed = pd.concat([X_scaled_df, y.reset_index(drop=True)], axis=1)

# Menyimpan Dataset Hasil Preprocessing
import os
os.makedirs('preprocessing', exist_ok=True)
df_preprocessed.to_csv('preprocessing/heart_preprocessed.csv', index=False)
print("Dataset preprocessed disimpan ke 'preprocessing/heart_preprocessed.csv'")

# Tampilkan 5 baris pertama hasil preprocessing
print("\nContoh hasil preprocessing:")
display(df_preprocessed.head())
```
